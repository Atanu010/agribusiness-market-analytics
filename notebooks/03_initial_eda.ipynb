### 1. KPI Calculation Methodology Testing
We verify the mathematical correctness of our KPI functions on benchmark series before executing on full empirical data in Week 3.

```python
# Test CAGR Calculation
start_p, end_p, years = 2000.0, 3100.0, 10
cagr_val = calculate_cagr(start_p, end_p, years)
print(f"CAGR ({years} years: {start_p} -> {end_p}): {cagr_val * 100:.2f}%")

# Test Volatility and CV on sample series
sample_prices = pd.Series([1850, 1900, 1875, 1920, 2100, 2050, 1980, 2150, 2200, 2180])
cv_val = calculate_coefficient_of_variation(sample_prices)
vol_val = calculate_price_volatility(sample_prices)
print(f"Coefficient of Variation (CV): {cv_val:.2f}%")
print(f"Historical Volatility (Log Returns Std): {vol_val:.4f}")
```

### 3. Forecast Evaluation Metric Suite Verification
Ensuring that MAE, RMSE, MAPE, and sMAPE are mathematically verified and robust to boundary cases.

```python
y_actual = np.array([2100.0, 2150.0, 2200.0, 2250.0, 2300.0])
y_predicted = np.array([2080.0, 2160.0, 2190.0, 2270.0, 2290.0])

metrics = evaluate_forecast(y_actual, y_predicted)
print("Forecast Evaluation Benchmark Results:")
for k, v in metrics.items():
    unit = "%" if "MAPE" in k else "units"
    print(f"  {k:6s}: {v:8.3f} {unit}")
```
